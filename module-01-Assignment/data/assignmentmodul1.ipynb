```python
import pandas as pd
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_parquet('./yellow_tripdata_2023-01.parquet')
```

```python
df.head()
```

```python
#Question - 1 --> Downloading the data
df.columns.tolist()
len(df.columns)
```

```python
#Question - 2 --> Computing duration
df['duration'] = df.tpep_dropoff_datetime - df.tpep_pickup_datetime
df['duration'] = df.duration.dt.total_seconds() / 60
```

```python
df.duration.mean()
```

```python
df.duration.std()
```

```python
#Question 3 --> Dropping outliers
len(df[(df.duration >= 1) & (df.duration <= 60)]) / len(df) * 100
```

```python
df = df[(df.duration >= 1) & (df.duration <= 60)].copy()
```

```python
# Question - 4 --> One-Hot encoding
categorical = ['PULocationID', 'DOLocationID']
df[categorical] = df[categorical].astype(str)
train_dicts = df[categorical].to_dict(orient='records')
dv = DictVectorizer()
X_train = dv.fit_transform(train_dicts)
print(f'Feature matrix size: {X_train.shape}')
```

```python
#Question - 5 --> Training a Model
target = 'duration'
y_train = df[target].values
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred = lr.predict(X_train)

print(f'Train RMSE Value: {mean_squared_error(y_train, y_pred, squared=False)}')
```

```python
# Question --6 --> apply model on validation set
categorical = ['PULocationID', 'DOLocationID']

def read_data(filename):
    df = pd.read_parquet(filename)

    df['duration'] = df.tpep_dropoff_datetime - df.tpep_pickup_datetime
    df.duration = df.duration.dt.total_seconds() / 60

    df = df[(df.duration >= 1) & (df.duration <= 60)].copy()

    df[categorical] = df[categorical].astype('str')
    
    return df
```

```python
df_val = read_data('./yellow_tripdata_2023-02.parquet')
```

```python
val_dicts = df_val[categorical].to_dict(orient='records')
X_val = dv.transform(val_dicts) 
y_val = df_val.duration.values
y_pred = lr.predict(X_val)
print(f'Val RMSE Value: {mean_squared_error(y_val, y_pred, squared=False)}')
```
